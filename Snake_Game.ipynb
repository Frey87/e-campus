## Step 2: Understanding the HTML5 Canvas API

```python
var canvas = document.createElement("canvas");
canvas.width = 520;
canvas.height = 520;
var ctx = canvas.getContext("2d");
document.body.appendChild(canvas);
```

## Step 3: Creating the Game Grid

```python
var COLS = 26, ROWS = 26;
var EMPTY = 0, SNAKE = 1, FRUIT = 2;

var grid = {
    width: COLS,
    height: ROWS,
    _grid: [],

    init: function(value) {
        for (var x = 0; x < this.width; x++) {
            this._grid[x] = [];
            for (var y = 0; y < this.height; y++) {
                this._grid[x][y] = value;
            }
        }
    },
    set: function(val, x, y) {
        this._grid[x][y] = val;
    },
    get: function(x, y) {
        return this._grid[x][y];
    }
};
```

## Step 4: Implementing the Snake

```python
var snake = {
    direction: RIGHT,
    _queue: [],
    last: null,

    init: function(d, x, y) {
        this.direction = d;
        this._queue = [];
        this.insert(x, y);
    },
    insert: function(x, y) {
        this._queue.unshift({x: x, y: y});
        this.last = this._queue[0];
    },
    remove: function() {
        return this._queue.pop();
    }
};
```

## Step 5: Adding Controls

```python
var keystate = {};
document.addEventListener("keydown", function(evt) {
    keystate[evt.keyCode] = true;
});
document.addEventListener("keyup", function(evt) {
    delete keystate[evt.keyCode];
});
```

## Step 6: Updating the Game Logic

```python
function update() {
    if (keystate[37] && snake.direction !== RIGHT) snake.direction = LEFT;
    if (keystate[38] && snake.direction !== DOWN) snake.direction = UP;
    if (keystate[39] && snake.direction !== LEFT) snake.direction = RIGHT;
    if (keystate[40] && snake.direction !== UP) snake.direction = DOWN;
}
```

## Step 7: Drawing the Game
